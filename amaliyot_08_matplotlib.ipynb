## 7. Mini-loyiha: tizimga kirish urinishlari jurnali

Server jurnalidan olingan ma'lumot: bir haftada har **soatda** nechta kirish urinishi bo'lgan
va ulardan nechtasi **muvaffaqiyatsiz** tugagan.

```python
r = np.random.default_rng(42)
vaqt = pd.date_range('2026-10-19', periods=7 * 24, freq='h')
soat = vaqt.hour
odatiy = np.where((soat >= 8) & (soat < 20), 120, 15) * np.where(vaqt.dayofweek >= 5, 0.4, 1)
urinish = r.poisson(odatiy)
xato = r.binomial(urinish, 0.05)
# Hujum: yakshanba 01:00–04:00
hujum = (vaqt.dayofweek == 6) & (soat >= 1) & (soat < 4)
urinish[hujum] += r.integers(600, 900, hujum.sum())
xato[hujum] = (urinish[hujum] * 0.9).astype(int)

jurnal = pd.DataFrame({'vaqt': vaqt, 'urinish': urinish, 'xato': xato})
jurnal['xato_ulush'] = jurnal['xato'] / jurnal['urinish']
jurnal.head()
```
