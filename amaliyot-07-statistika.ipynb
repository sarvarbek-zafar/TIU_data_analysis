```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

r = np.random.default_rng(7)

# --- 1) "Kelajak" MChJ: 9 xodimning oylik maoshi (mln so'm). Oxirgisi -- direktor.
maosh = pd.Series([3, 3, 3, 3.5, 4, 4.5, 5, 6, 58],
                  index=['Ali', 'Vali', 'Gani', 'Soli', 'Toli', 'Bobur', 'Zafar', 'Laylo', 'Direktor'])

# --- 2) 300 talaba: uyqu, telefon, davomat va imtihon bali
n = 300
uyqu = np.clip(r.normal(7, 1.1, n), 4, 10).round(1)
telefon = np.clip(r.gamma(2.2, 1.6, n), 0.3, 12).round(1)
davomat = np.clip(r.normal(82, 10, n), 40, 100).round()
ball = np.clip(18 + 3.2 * uyqu - 2.2 * telefon + 0.42 * davomat + r.normal(0, 6, n), 0, 100).round()
talabalar = pd.DataFrame({'guruh': r.choice(['A', 'B', 'C'], n), 'uyqu': uyqu,
                          'telefon': telefon, 'davomat': davomat, 'ball': ball})

# --- 3) "TezTaksi": 400 ta safar
m = 400
masofa = r.gamma(2, 3, m).round(1) + 0.5
narx = (8000 + 2600 * masofa + r.normal(0, 2500, m)).round(-2)
kutish = r.exponential(6, m).round(1)
kutish[r.choice(m, 6, replace=False)] = [48, 55, 62, 70, 85, 95]       # tirbandlik kunlari
reyting = r.choice([5, 4, 3, 2, 1], m, p=[.52, .25, .1, .05, .08])
taksi = pd.DataFrame({'masofa': masofa, 'narx': narx, 'kutish': kutish, 'reyting': reyting})

print(talabalar.shape, taksi.shape)
taksi.head()
```

## 4. Kim haqiqatan kuchliroq? — normal taqsimot va z-ball
