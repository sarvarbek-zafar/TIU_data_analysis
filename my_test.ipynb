```python
import sys
```

```python
def main():
    input_data=sys.stdin.read().split()
    if not input_data:
        return
    x=int(input_data[0])
    y=int(input_data[1])
    if (x+y)%3==0 and 2*x>=y and 2*y>=x:
        print("Yes")
    else:
        print("No")

if __name__== "__main__":
    main()
```
