## 3-bo'lim. `loc` va `iloc` bilan tanlash

### 📘 Namuna
- `iloc` — **o'rin raqami** bo'yicha (0 dan boshlanadi, oxirgi chegara kirmaydi);
- `loc` — **nom (yorliq)** yoki **mantiqiy shart** bo'yicha (oxirgi chegara kiradi).

```python
df = pd.DataFrame({'ism':   ['Ali', 'Vali', 'Gani', 'Soli', 'Toli'],
                   'guruh': ['A', 'B', 'A', 'B', 'B'],
                   'ball':  [70, 85, 92, 55, 61]})
df
```

```python
print(df.iloc[1, 2])                 # 2-qator, 3-ustun: 85
print(df.iloc[-2:])                  # oxirgi 2 qator
print(df.iloc[::2])                  # har ikkinchi qator (0, 2, 4)

print(df.loc[df['guruh'] == 'A', ['ism', 'ball']])   # shart + ustunlar

t = df.set_index('ism')              # ismni indeksga aylantiramiz
print(t.loc['Gani', 'ball'])         # 92

df.loc[df['ism'] == 'Soli', 'ball'] = 60   # loc bilan qiymatni o'zgartirish
print(df['ball'].mean())             # 73.6
```
